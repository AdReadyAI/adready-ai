# Inspect frame-sampling results
Loads `report.json` produced by `evaluate_frame_sampling.py` and previews selected frames.

```python
import json
import os

REPORT_PATH = "tmp/report.json"

with open(REPORT_PATH) as f:
    report = json.load(f)
```

```python
from PIL import Image
import matplotlib.pyplot as plt

THUMB_SIZE = (320, 320)

def plotSelectedFrames(frames: list):
    print(f"{len(frames)} frames selected")
    
    cols = 4
    rows = (len(frames) + cols - 1) // cols
    cols = 4
    rows = (len(frames) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(4 * cols, 3 * rows), dpi=80)

    for ax, frame in zip(axes.flat, frames):
        with Image.open(frame["path"]) as img:
            img.thumbnail(THUMB_SIZE)
            ax.imshow(img)
        ax.set_title(f"idx={frame['index']} t={frame['timestamp']:.2f}s\n{frame['tags']}", fontsize=8)
        ax.axis("off")

    for ax in axes.flat[len(frames):]:
        ax.axis("off")

    plt.tight_layout()
    plt.show()
    plt.close(fig)
```

```python
def getProbeFrames(name: str):
    return [frame for frame in report["frames"] if name in frame["tags"]]
```
